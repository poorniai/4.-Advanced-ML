```python
import pandas as pd
import SelectKHelper
import SelectK_Classification_Algorithm
```

```python
dataset = pd.read_csv("prep.csv")
dataset
df=dataset
```

```python
df=pd.get_dummies(df,dtype=int,drop_first=True)
```

```python
indep_X=df.drop('classification_yes',axis=1)
dep_Y=df['classification_yes']
```

```python
def selectKAlgorithm(n):
    kbest,selected_columns= SelectKHelper.selectkbest(indep_X,dep_Y,n)
    X_train,X_test,y_train,y_test=SelectKHelper.standard_split(kbest,dep_Y)
    acclog=[]
    accsvmline=[]
    accsvmnl=[]
    navie=[]
    knn=[]
    decision=[]
    random=[]
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.logistic(X_train,X_test,y_train,y_test)
    acclog.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.svm_linear(X_train,X_test,y_train,y_test)
    accsvmline.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.svm_NL(X_train,X_test,y_train,y_test)
    accsvmnl.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.Navie(X_train,X_test,y_train,y_test)
    navie.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.KNN(X_train,X_test,y_train,y_test)
    knn.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.Decision(X_train,X_test,y_train,y_test)
    decision.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = SelectK_Classification_Algorithm.Random(X_train,X_test,y_train,y_test)
    random.append(Accuracy)
    selected_table = pd.DataFrame([[n] + list(selected_columns)],columns=['kvalue'] + [f'Feature_{i+1}' for i in range(len(selected_columns))])
    result=SelectK_Classification_Algorithm.Select_Classification(n,acclog,accsvmline,accsvmnl,navie,knn,decision,random)
    return selected_table, result
```

```python
selected_table,result = selectKAlgorithm(4)
print(selected_table.to_string(index=False))
print("\n")
print(result)
```
