```python
import pandas as pd
import RFE_Helper
import RFE_Classification_Algorithm
```

```python
dataset = pd.read_csv("prep.csv")
df = dataset
```

```python
df=pd.get_dummies(df,dtype=int,drop_first=True)
df
```

```python
indep_X=df.drop('classification_yes',axis=1)
dep_Y=df['classification_yes']
```

```python
def rfe_classification_algo(indep_X,dep_Y,n):
    rfeList = RFE_Helper.rfeFeature(indep_X,dep_Y,n)
    acclog=[]
    accsvmline=[]
    accsvmnl=[]
    accnav=[]
    accknn=[]
    accdes=[]
    accrf=[]
    for i in rfeList:   
        X_train, X_test, y_train, y_test=RFE_Helper.standard_split(i,dep_Y)   
        
            
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.logistic(X_train,X_test,y_train,y_test)
        acclog.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.svm_linear(X_train,X_test,y_train,y_test) 
        accsvmline.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.svm_NL(X_train,X_test,y_train,y_test)  
        accsvmnl.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.KNN(X_train,X_test,y_train,y_test)  
        accknn.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.Navie(X_train,X_test,y_train,y_test)
        accnav.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.Decision(X_train,X_test,y_train,y_test)  
        accdes.append(Accuracy)
        
        classifier,Accuracy,report,X_test,y_test,cm=RFE_Classification_Algorithm.Random(X_train,X_test,y_train,y_test)  
        accrf.append(Accuracy)
    
    result=RFE_Classification_Algorithm.rfe_classification(acclog,accsvmline,accsvmnl,accknn,accnav,accdes,accrf)

    return result
    
```

```python
result = rfe_classification_algo(indep_X,dep_Y,3)
result
```
