```python
import pandas as pd
import RFE_Helper
import RFE_Regression_Algorithm
```

```python
dataset = pd.read_csv("prep.csv")
df = dataset
```

```python
df=pd.get_dummies(df,dtype=int,drop_first=True)
df
```

```python
indep_X=df.drop('classification_yes',axis=1)
dep_Y=df['classification_yes']
```

```python
def rfe_regression_algo(indep_X,dep_Y,n):
    rfelist=RFE_Regression_Algorithm.rfeFeatureReg(indep_X,dep_Y,n)
    acclin=[]
    accsvml=[]
    accsvmnl=[]
    accdes=[]
    accrf=[]


    for i in rfelist:   
        X_train, X_test, y_train, y_test=RFE_Helper.standard_split(i,dep_Y)  
        r2_lin=RFE_Regression_Algorithm.Linear(X_train,y_train,X_test,y_test)
        acclin.append(r2_lin)
        
        r2_sl=RFE_Regression_Algorithm.svm_linear(X_train,y_train,X_test,y_test)    
        accsvml.append(r2_sl)
        
        r2_NL=RFE_Regression_Algorithm.svm_NL(X_train,y_train,X_test,y_test)
        accsvmnl.append(r2_NL)
        
        r2_d=RFE_Regression_Algorithm.Decision(X_train,y_train,X_test,y_test)
        accdes.append(r2_d)
        
        r2_r=RFE_Regression_Algorithm.random(X_train,y_train,X_test,y_test)
        accrf.append(r2_r)
        
        
    result=RFE_Regression_Algorithm.rfe_regression(acclin,accsvml,accdes,accrf)
    
    return result  
    
    
    
```

```python

result=rfe_regression_algo(indep_X,dep_Y,3)
result
```

```python
result=rfe_regression_algo(indep_X,dep_Y,4)
result
```
