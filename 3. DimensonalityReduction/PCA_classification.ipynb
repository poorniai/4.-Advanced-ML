```python
import pandas as pd
import DimensionalityReductionHelper
import Classification_Algorithm
```

```python
dataset = pd.read_csv("prep.csv")
dataset
df=dataset
```

```python
df=pd.get_dummies(df,dtype=int,drop_first=True)
```

```python
indep_X=df.drop('classification_yes',axis=1)
dep_Y=df['classification_yes']
```

```python
def pcaAlgorithm(n):
    X_train_pca = DimensionalityReductionHelper.pca(indep_X,n)
    X_train,X_test,y_train,y_test=DimensionalityReductionHelper.standard_split(X_train_pca,dep_Y)
    acclog=[]
    accsvmline=[]
    accsvmnl=[]
    navie=[]
    knn=[]
    decision=[]
    random=[]
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.logistic(X_train,X_test,y_train,y_test)
    acclog.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.svm_linear(X_train,X_test,y_train,y_test)
    accsvmline.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.svm_NL(X_train,X_test,y_train,y_test)
    accsvmnl.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.Navie(X_train,X_test,y_train,y_test)
    navie.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.KNN(X_train,X_test,y_train,y_test)
    knn.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.Decision(X_train,X_test,y_train,y_test)
    decision.append(Accuracy)
    classifier,Accuracy,report,X_test,y_test,cm = Classification_Algorithm.Random(X_train,X_test,y_train,y_test)
    random.append(Accuracy)
    result=Classification_Algorithm.Select_Classification(n,'PCA-Value',acclog,accsvmline,accsvmnl,navie,knn,decision,random)
    return result,X_train_pca
```

```python

result,X_train_pca = pcaAlgorithm(3)
print(X_train_pca)
print("\n")
print(X_train_pca.shape)
print("\n")
print(result.to_string(index=False))
```
