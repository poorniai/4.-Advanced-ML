```python
import pandas as pd
import SelectKHelper
import SelectK_Regression_Algorithm
```

```python
dataset = pd.read_csv("prep.csv")
dataset
df=dataset
```

```python
df=pd.get_dummies(df,dtype=int,drop_first=True)
```

```python
indep_X=df.drop('classification_yes',axis=1)
dep_Y=df['classification_yes']
```

```python
def selectKAlgorithm(n):
     kbest,selected_columns= SelectKHelper.selectkbest(indep_X,dep_Y,n)
     X_train,X_test,y_train,y_test=SelectKHelper.standard_split(kbest,dep_Y)
     
     acclin=[]
     accsvml=[]
     accsvmnl=[]
     accdes=[]
     accrf=[]
     r2_lin=SelectK_Regression_Algorithm.Linear(X_train,y_train,X_test,y_test)
     acclin.append(r2_lin)
        
     r2_sl=SelectK_Regression_Algorithm.svm_linear(X_train,y_train,X_test,y_test)    
     accsvml.append(r2_sl)
        
     r2_NL=SelectK_Regression_Algorithm.svm_NL(X_train,y_train,X_test,y_test)
     accsvmnl.append(r2_NL)
        
     r2_d=SelectK_Regression_Algorithm.Decision(X_train,y_train,X_test,y_test)
     accdes.append(r2_d)
         
     r2_r=SelectK_Regression_Algorithm.random(X_train,y_train,X_test,y_test)
     accrf.append(r2_r)
    
     result=SelectK_Regression_Algorithm.selectk_regression(acclin,accsvml,accsvmnl,accdes,accrf)
     selected_table = pd.DataFrame([[n] + list(selected_columns)],columns=['kvalue'] + [f'Feature_{i+1}' for i in range(len(selected_columns))])

     return selected_table,result
```

```python
result,selected_table=selectKAlgorithm(4)
print(selected_table.to_string(index=False))
print("\n")
print(result)
```
